```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("../STOCK MARKET/AXISBANK.csv")
data.head()
```

```python
data['Trades'].fillna(data['Trades'].median(),inplace=True)
data['Deliverable Volume'].fillna(data['Deliverable Volume'].median(),inplace=True)
data['%Deliverble'].fillna(data['%Deliverble'].median(),inplace=True)
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
data.describe()
```

```python
data['Date']=pd.to_datetime(data['Date'])
```

```python
data.head()
```

```python
plt.figure(figsize=(25,10))
plt.plot(data['Date'],data['Close'])
plt.ylabel("Stock Price")
plt.xlabel("Year")
plt.title("Stock Price Over Time")
```

```python
data['MA50'] = data['Close'].rolling(window=50).mean()
data['MA200'] = data['Close'].rolling(window=200).mean()

plt.figure(figsize=(25,10))
plt.plot(data['Date'],data['Close'],label='Close Price')
plt.plot(data['Date'],data['MA50'],label='50-day MA',color='red')
plt.plot(data['Date'],data['MA200'],label='200-day MA',color='green')
plt.ylabel("Stock Price")
plt.xlabel("Year")
plt.legend()
plt.title("Stock Price With Moving Average")
```

```python
plt.figure(figsize=(12,6))
sns.heatmap(data.corr(),annot=True)
```

```python
data['daily_return']=data['Close'].pct_change()
```

```python
plt.hist(data['daily_return'].dropna(),bins=50,color='purple',alpha=0.7)
plt.xlabel('Daily Return')
plt.ylabel('Frequency')
plt.title('Stock Daily Returns Distribution')
plt.show()
```

```python
data['Price_change'] = data['Close'] - data['Open']
data['High-low'] = data['High'] - data['Low']
data['Volatility'] = data['Close'].rolling(window=10).std()
```

```python
data.head()
```

```python
features = ['Open','High','Low','Volume','VWAP']
target = 'Close'
data.dropna(inplace=True)
```

```python
from sklearn.preprocessing import MinMaxScaler
mm = MinMaxScaler()
data[features] = mm.fit_transform(data[features])
data.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(data[features],data[target],test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
lr.score(x_test,y_test),lr.score(x_train,y_train)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
```

```python
print(f'Mean Absolute Error: {mae}')
print(f'Mean Squared Error: {mse}')
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
plt.figure(figsize=(12,6))
plt.plot(y_test.values,label="Actual Prices")
plt.plot(y_pred,label="Predicted Prices", color='g',alpha=0.5, linestyle='dotted')
plt.legend()
plt.title("Stock Price Prediction using Linear Regression")
plt.show()
```
